```python
import pandas as pd
from google.colab import files
import io

# Upload the Excel file
uploaded = files.upload()

if not uploaded:
    print("No file uploaded. Please upload your Excel file.")
else:
    # Get the file name from the uploaded dictionary (assuming a single file upload)
    file_path = next(iter(uploaded))

    # Specify the name of the main sheet you want to read
    # If not specified, pandas reads the first sheet by default.
    sheet_name = '' # Replace with your actual sheet name, or remove if you want the first sheet

    try:
        # Load the Excel file into a pandas DataFrame from the uploaded content
        # Use header=2 to set the third row (0-indexed) as the header
        df = pd.read_excel(io.BytesIO(uploaded[file_path]), header=2)
        print(f"Successfully loaded data from '{file_path}'")

        # Display the first 5 rows of the DataFrame
        print("First 5 rows of the DataFrame with corrected headers:")
        display(df.head())

    except ValueError as e:
        print(f"Error: {e}. Please check the sheet name or if the file is a valid Excel file.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```

```python
columns_to_keep = [
    'Item No',
    'Description',
    'Description 2',
    'UPC Code',
    'Count unit',
    'Alt Unit 1',
    'Convert 0 to 1 factor',
    'Convert 1 to 2 factor',
    'Length',
    'Width',
    'Height',
    'Unit weight',
    'Hazardous'
]

# Filter the DataFrame to keep only the specified columns
df = df[columns_to_keep]

print("DataFrame after keeping only specified columns:")
display(df.head())
```

```python

df['Hazardous'] = df['Hazardous'].apply(lambda x: 'Hazardous' if x == True else '')


df.loc[df['Convert 0 to 1 factor'] == 0, 'Convert 0 to 1 factor'] = 1

df.loc[df['Convert 1 to 2 factor'] == 0, 'Convert 1 to 2 factor'] = 1

display(df.head())
```

```python
new_df = pd.DataFrame({
    'SKU': df['Item No'],
    'Description': df['Description'],
    'Description2': df['Description 2'],
    'Min': '',
    'Max': '',
    'Cycle Count': '',
    'ReorderQty': '',
    'Inventory Method': '',
    'Temperature': '',
    'Cost': '',
    'UPC': df['UPC Code'],
    'Track Lot': True,
    'Track Serial': False,
    'Track ExpDate': False,
    'Primary Unit of Measure': df['Count unit'],
    'Packaging Unit': df['Alt Unit 1'],
    'Packing UOM Qty': df['Convert 0 to 1 factor'],
    'Length': df['Length'],
    'Width': df['Width'],
    'Height': df['Height'],
    'Weight': df['Unit weight'],
    'Qualifiers': '',
    'Storage Setup': '',
    'Variable Setup': '',
    'NMFC#': '',
    'Lot Number Required': False,
    'Serial Number Required': False,
    'Serial Number': False,
    'Exp Date Req': False,
    'Enable Cost': False,
    'Cost Required': False,
    'IsHazMat': df['Hazardous'],
    'HazMatID': '',
    'HazMatShippingName': '',
    'HazMatHazardClass': '',
    'HazMatPackingGroup': '',
    'HazMatFlashPoint': '',
    'HazMatLabelCode': '',
    'HazMatFlag': '',
    'Image URL': '',
    'StorageCountScriptTemplateID': '',
    'StorageRates': '',
    'OutboundMobileSerializationBehavior': '',
    'Price': '',
    'TotalQty': (df['Convert 0 to 1 factor'] * df['Convert 1 to 2 factor']),
    'UnitType': 'PALLET'
})

print("New DataFrame created with specified columns:")
display(new_df.head())
```

```python
from google.colab import files

# Define the output Excel file name
output_filename = 'new_dataset.xlsx'

# Save the new DataFrame to an Excel file
new_df.to_excel(output_filename, index=False)

# Download the file
files.download(output_filename)

print(f"DataFrame successfully saved to '{output_filename}' and ready for download.")
```
